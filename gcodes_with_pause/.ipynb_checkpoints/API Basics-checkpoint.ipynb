```python
import time
import bambulabs_api as bl

IP = '172.16.97.160'
SERIAL = '01P00C641402808'
ACCESS_CODE = '42117904'

if __name__ == '__main__':
    print('Starting bambulabs_api example')
    print('Connecting to Bambulabs 3D printer')
    print(f'IP: {IP}')
    print(f'Serial: {SERIAL}')
    print(f'Access Code: {ACCESS_CODE}')

    # Create a new instance of the API
    printer = bl.Printer(IP, ACCESS_CODE, SERIAL)

    # Connect to the Bambulabs 3D printer
    printer.connect()

    time.sleep(2)

    # Get the printer status
    status = printer.get_state()
    print(f'Printer status: {status}')

    # Turn the light off
    printer.turn_light_off()

    time.sleep(5)

    # Turn the light on
    printer.turn_light_on()

    # Disconnect from the Bambulabs 3D printer
    printer.disconnect()
```

```python
import time
import bambulabs_api as bl

IP = '172.16.97.160'
SERIAL = '01P00C641402808'
ACCESS_CODE = '42117904'

# Point directly to your .gcode.3mf file
INPUT_3MF_PATH = 'modified.gcode.3mf'
UPLOAD_FILE_NAME = 'modified.3mf'
PLATE_INDEX = 1  # Standard plate number to print (usually 1)

if __name__ == '__main__':
    print('Connecting to Bambu Lab printer...')
    printer = bl.Printer(IP, ACCESS_CODE, SERIAL)
    printer.connect()

    time.sleep(5)

    # Open the .gcode.3mf file in binary mode
    with open(INPUT_3MF_PATH, "rb") as file_stream:
        print(f"Uploading {INPUT_3MF_PATH} to printer...")
        result = printer.upload_file(file_stream, UPLOAD_FILE_NAME)

    # FTP server usually returns an FTP code containing '226' (Transfer Complete)
    if "226" not in str(result):
        print(f"Error uploading file to printer: {result}")
    else:
        print("Upload successful. Sending start print command...")
        # start_print parameters: (target_file_on_printer, plate_index)
        printer.start_print(UPLOAD_FILE_NAME, PLATE_INDEX)
        print("Start print command sent successfully.")
```
